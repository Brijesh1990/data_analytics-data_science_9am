```python
# swap two numbers with using third variables 
a=10
b=20
temp=0
# applied logic ....
temp=a # 10
a=b    # 20
b=temp # 10

print("after swap a is :",a)
print("after swap b is :",b)
```

```python
# w.a.p to swap two numbers with using third variables take input from users
# input() is an method that can be take input from users 
a=int(input("Enter a values :"))
b=int(input("Enter b values :"))
temp=0
temp=a
a=b
b=temp
print("after swap a is :",a)
print("after swap b is :",b)
```

```python
# w.a.p to swap two numbers without using third variables take input from users
# input() is an method that can be take input from users 
a=int(input("Enter a values :"))
b=int(input("Enter b values :"))
# logic to swap two numbers without using third variables  
a=a+b #30
b=a-b #10
a=a-b #20 
print("after swap a is :",a)
print("after swap b is :",b)
```

```python
# w.a .p to calculate  simple interest take input from users 
p=int(input("Enter principle ammount :"))
n=int(input("Enter Numbers of Time  :"))
r=int(input("Enter ROI ammount :"))
# formula 
si=p*n*r/100
print("The interest you have to paid :",si)
```

```python
# find the square root of numbers 
# using sqrt √25 = 5
import math
number=int(input("Enter a numbers :"))
res=math.sqrt(number);
print(res)
```

```python
# w.a .p to calculate  compund interest take input from users 
p=int(input("Enter principle ammount :"))
n=int(input("Enter Numbers of Time  :"))
r=int(input("Enter ROI ammount :"))
# formula 
ci=p*(pow((1+r/100),n))
print("The compund interest you have to paid :",ci)
```

```python
# area of circle ... 
# take input from users 
# formula of area of circle 𝐴𝑟𝑒𝑎=π*r²
r=int(input("Enter radius of circle : "))
a=3.14*pow(r,2)
print(a)
```
